```python
import pandas as pd
```

```python
data = {
    "Name": ["Ram", "Shyam", "Ghanshyam", "Dhanshyam", "Aditi", "Jagdish", "Raj", "Aman"],
    "Age": [28, 34, 22, 30, 29, 40, 25, 32],
    "Salary": [50000, 60000, 45000, 52000, 49000, 70000, 48000, 58000],
    "Performance Score": [85, 90, 78, 92, 88, 95, 80, 89]
}

df = pd.DataFrame(data)
```

```python
print("Sample Data")
print(df)
```

```python
#Select Single column
print("Names :(Single columns returns series)")
print(df["Name"])
```

```python
#Select Multiple Columns
print("for selecting multiple columns")
mul_col = df[["Name","Age", "Salary"]]
print(mul_col)
```

```python

#Select fro filtering  -------------- BOTH CONDITION HAVE TRUE ------------------
print("here data after filter") 
filtere_01 = df[(df['Age'] > 30) & (df['Salary'] > 50000 )]
print(filtere_01)
```

```python
#using condition OR ------------------- WORK IF ONLY SINGLE CONDITON HAVE TRUE ----------------
print("performance score > 90  or age > 30 ")
filter_02 = df[(df["Age"] > 35 )| (df["Performance Score"] > 90)]
print(filter_02)
```

```python
#  Adding columns straight forword

df["Bonus"] = df["Salary"]* 0.1
print(df)
```

```python
#Using insert for adding columns

df.insert(0,"Employee_id", range(101, 101+len(df))) # 0 = location --- yaniki hamara column number kaunsa rahega 
print(df)
```

```python
# Update any value in Dataset 
df.loc[0, "Salary"] = 55000   # we update RAM's salary
print(df)
```

```python
# Update more than 1 column

df["Salary"] = df["Salary"] * 1.05
print(df)
```

```python
# Drop columns
df.drop(columns=["Performance Score", "Age"], inplace=True)
print(df)
```

```python
import pandas as pd
```

```python
df_customer = pd.DataFrame({
    "Name": ["Mohan", "Ajay" , "Shree" , "Hariom"],
    "id": [101,102,103,104]
})
```

```python
df_orders = pd.DataFrame({
    "id":[103,102, 104],
    "amount": [250,450,750]
}
)
```

```python
df_merge = pd.merge(df_customer, df_orders, on="id", how="inner")
print(df_merge)
```

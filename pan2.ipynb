```python
import pandas as pd
```

```python
data = {
    "Name": ['Ram', None, 'Ghanshyam', 'Dhanshyam', 'Aditi', 'Jagdish', 'Raj', 'Simran'],
    "Age": [28, None, 22, 30, 29, 40, 25, 32],
    "Salary": [50000, None, 45000, 52000, 49000, 70000, 48000, 58000],
    "Performance_Score": [85, None, 78, 92, 88, 95, 80, 89]
}

df = pd.DataFrame(data)
print(df)
```

```python
#missing value

print(df.isnull().sum())  # isnull() returns bollen value
```

```python
# drop missing values 
#axis 0 = rows ki missing value ko hatana 
#axis = 1 column ki missing value ko hatana
df.dropna(inplace=True)
print(df)
```

```python
#replace missing value with new
#fillana(value, inpace=true)

# df.fillna(0, inplace=True)
# print(df)

df["Age"] = df["Age"].astype(float)

df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)
```

```python
fdata = {
    "Time": [1,2,3,4,5],
    "Value": [10, None, 30, None, 50]
}

fdf = pd.DataFrame(fdata)

print(fdf)
```

```python
fdf["Value"] = fdf["Value"].interpolate(method="linear")

print(fdf)
```

```python
#shorting with multiple columns
kdata = {
    "Name": ['Arun', 'Varun', 'Karun','Ajay','Sanjay','Ram'],
    "Age": [28, 34, 22, 34, 22, 22],
    "Salary": [10000, 20000, 30000, 45000, 25000, 87000]
}
kdf = pd.DataFrame(kdata)
print(kdf)
```

```python
#shorting 

kdf.sort_values(by=["Age", "Salary"], ascending=[True, False])
print(kdf)
```

```python
sum_sal = kdf["Salary"].sum()
print(sum_sal)
print(kdf)
```

```python
#Grouping lya fir group by 
grouped_sal = kdf.groupby("Age")["Salary"].sum()
print(grouped_sal)
print(kdf)
```
